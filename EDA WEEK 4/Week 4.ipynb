```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Cleaned ShopifyData.csv")
df
```

```python
df.set_index(df["date"])
```

```python
plt.figure(figsize = (12,6))

plt.plot(df.index,df["open"],label="Open")
```

```python
plt.figure(figsize = (12,6))

plt.plot(df.index,df["high"],label="high")
```

```python
plt.figure(figsize = (12,6))

plt.plot(df.index,df["low"],label="low")
```

```python
plt.figure(figsize = (12,6))

plt.plot(df.index,df["close"],label="close")
```

```python
plt.figure(figsize=(12,6))

plt.plot(df.index, df["open"],label="open",color = "blue")
plt.plot(df.index, df["high"],label="high",color = "green")
plt.plot(df.index, df["low"],label="low",color = "purple")
plt.plot(df.index, df["close"],label="close",color = "yellow")

plt.title("Shopify Stock Prices Over Time")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.plot(df.index, df["volume"])
    
plt.title("Shopify Stock Prices Over Time")
plt.xlabel("Date")
plt.ylabel("Price")
plt.xticks(rotation=45)
plt.show()
```

```python
df["SMA_20"] = df["close"].rolling(window=20).mean()

df["SMA_50"] = df["close"].rolling(window=50).mean()

print(df[["close" , "SMA_20", "SMA_50"]].head(55))
```

```python
plt.figure(figsize=(12,6))

plt.plot(df.index, df["close"],label="Close Price")
plt.plot(df.index, df["SMA_20"],label="20-DAY SMA")
plt.plot(df.index, df["SMA_50"],label="50-DAY SMA ")

plt.title("Shopify Closing Price with Moving Averages")
plt.xlabel("Date")
plt.ylabel("Price")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["volume"], df["Daily_Price_Change"])

plt.title("Trading volume vs Daily Price Change")
plt.xlabel("Trading volume")
plt.ylabel("Daily Price Change")

plt.show()
```

```python
print("Mean Return:", df["Daily_Return"].mean())
print("Variance:", df["Daily_Return"].var())
print("Standard Deviation:", df["Daily_Return"].std())
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Daily_Return"],bins=20)

plt.title("Distribution of shopify Daily Returns")
plt.xlabel("Daily Return (%)")
plt.ylabel("Frequency")

plt.show()           
```

```python
plt.figure(figsize=(8,5))

sns.kdeplot(df["Daily_Return"],fill=True)

plt.title("KDE Plot of Shopify Daily Return")
plt.xlabel("Daily Return (%)")

plt.show()
```

```python
threshold = df["Daily_Return"].abs().quantile(0.90)

high_volatility = df[
    df["Daily_Return"].abs()>=threshold
    ]

print(high_volatility[
    ["close","Daily_Return"]
    ])
```
